**11.1 TODO, INTERACTIVE.** Fill in the skeleton below. It reads a gross income and a number of dependents, then prints the income tax, following the slides' `taxform` example.

Tax rule: taxable income is the gross income minus 10000 per dependent. Tax is 20 percent of the taxable income, and never below zero.

````python
"""
Program: Income Tax Calculator
Purpose: Calculate income tax based on gross income and number of dependents.
Input: Gross income and number of dependents.
Output: Income tax.
"""

# TODO: constants
TAX_RATE = 0.20
EXEMPTION_PER_DEPENDENT = 10000

# input
grossIncome = float(input("Enter the gross income: "))
dependents = int(input("Enter the number of dependents: "))

# processing
taxableIncome = grossIncome - (dependents * EXEMPTION_PER_DEPENDENT)   # TODO
tax =  max(0, taxableIncome * TAX_RATE)            # TODO

# output
print("The income tax is $" + str(tax))
````
