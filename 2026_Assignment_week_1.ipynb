## 4️⃣ Type Conversion: `int()` and `float()`

Remember: `input()` always returns a **string**. When you need to do math with numbers from user input, you must convert them!

- `int()` - converts to an integer (whole number)
- `float()` - converts to a floating-point number (decimal number)

### Syntax:
```python
number = int(input("Enter a number: "))
decimal = float(input("Enter a decimal: "))
```

### Example 4.1: Why Type Conversion Matters

````python
# Without conversion - this will concatenate (join) strings!
num1 = input("Enter first number: ")
num2 = input("Enter second number: ")
result = num1 + num2
print("Result:", result)  # Try entering 5 and 3 - you'll get "53" instead of 8!
````

````python
# With int() conversion - now it does math!
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
result = num1 + num2
print("Result:", result)  # Now 5 and 3 will give you 8!
````

### Example 4.2: Adding Two Numbers (from Class)

````python
# This is the example from our slides!
first = int(input("Enter the first number: "))
second = int(input("Enter the second number: "))
print("The sum is", first + second)
````

### Example 4.3: Using float() for Decimals

````python
# Using float() for decimal numbers
price = float(input("Enter the price: "))
tax_rate = 0.06
total = price + (price * tax_rate)
print("Total with tax:", total)
````

### Exercise 4.1: Temperature Converter

Write a program that converts Celsius to Fahrenheit.

Formula: `fahrenheit = (celsius * 9/5) + 32`

Your program should:

3. Calculate the Fahrenheit equivalent
4. Print the result

### Exercise 4.2: Simple Calculator

Create a simple calculator program that:

1. Asks the user for two numbers

   - Their sum
   - Their difference (first - second)
   - Their product (multiplication)
   - Their quotient (first ÷ second)

````python
# Exercise 4.2: Write your code here

# 1. Get two numbers from the user
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

# 2. Calculate sum, difference, product, and quotient
sum_result = num1 + num2
difference = num1 - num2
product = num1 * num2
quotient = num1 / num2 if num2 != 0 else "Error: Division by zero is not allowed."

# 3. Print all results
print("Sum:", sum_result)
print("Difference:", difference)
print("Product:", product)
print("Quotient:", quotient)
````

## 5️⃣ Understanding Syntax Errors

Syntax errors occur when Python doesn't understand your code because it doesn't follow Python's rules (syntax).

Common syntax errors include:
- Misspelled variable names
- Missing parentheses or quotation marks
- Incorrect indentation
- Incomplete expressions

### Example 5.1: Common Syntax Errors

**Run each cell below to see the error messages. Then fix the code and run it again!**

````python
# Error 1: Misspelled variable name
length = 44
print(length)  # This will cause a NameError - can you fix it?
````

````python
# Error 2: Missing closing quotation mark
print("Hello, world!")  # This will cause a SyntaxError - can you fix it?
````

````python
# Error 3: Incomplete expression
result = 3 + 5  # This will cause a SyntaxError - can you fix it?
print (result)
````

````python
# Error 4: Missing parenthesis
print("Python is fun")  # This will cause a SyntaxError - can you fix it?
````

### Tips for Debugging:

1. **Read the error message carefully** - it tells you what's wrong and where!
2. **Check the line number** mentioned in the error
3. **Look for common mistakes**: missing quotes, parentheses, or misspellings
4. **Run your code frequently** to catch errors early
5. **Don't panic!** Errors are normal and help you learn

### Challenge 1: Area Calculator

Write a program that calculates the area of a rectangle.

Your program should:
1. Ask the user for the length
2. Ask the user for the width
3. Calculate the area (length × width)
4. Calculate the perimeter (2 × length + 2 × width)
5. Print both results with descriptive messages

### Challenge 2: Shopping Bill Calculator

Create a program that calculates a shopping bill.

Your program should:
1. Ask for the price of three items
2. Calculate the subtotal (sum of all prices)
3. Calculate the sales tax (6% of subtotal)
4. Calculate the final total (subtotal + tax)
5. Print the subtotal, tax, and final total

````python
# Challenge 2: Write your code here
# Shopping Bill Calculator

# Ask for the price of three items
item1 = float(input("Enter the price of item 1: "))
item2 = float(input("Enter the price of item 2: "))
item3 = float(input("Enter the price of item 3: "))

# Calculate subtotal
subtotal = item1 + item2 + item3

# Calculate sales tax (6%)
tax = subtotal * 0.06

# Calculate final total
final_total = subtotal + tax

# Print the results
print("\n--- Shopping Bill ---")
print(f"Subtotal: ${subtotal:.2f}")
print(f"Sales Tax (6%): ${tax:.2f}")
print(f"Final Total: ${final_total:.2f}")
````
